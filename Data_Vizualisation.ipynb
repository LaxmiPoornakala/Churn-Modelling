```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
churn_data = pd.read_csv('/content/Churn_Modelling.csv')
churn_data.head()
```

```python
churn_data.shape
```

```python
churn_data.info()
```

```python
# Vizualization of Credit Score
plt.figure(figsize=(5, 3))

sns.histplot(data=churn_data, x='CreditScore', bins=30, kde=True, color='purple')
plt.title('Credit Score distribution')
plt.xlabel('Credit Score')
plt.ylabel('Frequency')
plt.show()
```

```python
# Vizualization of Geography
plt.figure(figsize=(5, 3))
sns.histplot(data=churn_data, x='Geography', bins=30, kde=True, color='skyblue')
plt.title('Distribution of Geography')
plt.xlabel('Geography')
plt.ylabel('Frequency')
plt.show()
```

```python
# Vizualization of Age
plt.figure(figsize=(5, 3))

sns.histplot(data=churn_data, x='Age', bins=30, kde=True, color='blue')
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
# Vizualization of Tenure
plt.figure(figsize=(5, 3))
sns.histplot(data=churn_data, x='Tenure', bins=30, kde=True, color='skyblue')
plt.title('Distribution of Tenure')
plt.xlabel('Tenure')
plt.ylabel('Frequency')
plt.show()
```

```python
# Vizualization of Balance
plt.figure(figsize=(5, 3))

sns.histplot(data=churn_data, x='Balance', bins=30, kde=True, color='green')
plt.title('Balance value distribution')
plt.xlabel('Balance')
plt.ylabel('Frequency')
plt.show()
```

```python
# Vizualization of Number Of Products
plt.figure(figsize=(5, 3))

churn_data['NumOfProducts'].value_counts().plot(kind='bar', color='pink')
plt.title('Distribution of Number of Products')
plt.xlabel('Number of Products')
plt.ylabel('Frequency')
plt.show()
```

```python
# Vizualization of Having Credit Card
plt.figure(figsize=(5, 3))

churn_data['HasCrCard'].value_counts().plot(kind='bar', color='yellow')
plt.title('Credit Card Holders')
plt.xlabel('Has Credit Card')
plt.ylabel('Frequency')
plt.xticks(ticks=[0, 1], labels=['No', 'Yes'], rotation=0)
plt.show()
```

```python
# Vizualization of Active members

plt.figure(figsize=(5, 3))
churn_data['IsActiveMember'].value_counts().plot(kind='bar', color='purple')
plt.title('Distribution of Active Members')
plt.xlabel('Is Active Member')
plt.ylabel('Frequency')
plt.xticks(ticks=[0, 1], labels=['No', 'Yes'], rotation=0)
plt.show()
```

```python
# Vizualization of Exited Customers
plt.figure(figsize=(5, 3))

churn_data['Exited'].value_counts().plot(kind='bar', color='red')
plt.title('Exited Customers')
plt.xlabel('Exited')
plt.ylabel('Frequency')
plt.xticks(ticks=[0, 1], labels=['No', 'Yes'], rotation=0)
plt.show()
```
